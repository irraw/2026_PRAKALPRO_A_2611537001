## 2.1 Dasar Teori

<div style="font-family:'Times New Roman', Times, serif; font-size: 18px;">

Operator merupakan simbol atau tanda yang digunakan dalam pemrograman untuk melakukan suatu operasi terhadap data atau variabel. Dalam bahasa pemrograman Python, operator digunakan untuk melakukan perhitungan, memberikan nilai, membandingkan data, serta menentukan kondisi tertentu. Penggunaan operator merupakan salah satu dasar penting dalam pembuatan program karena dapat membantu komputer melakukan proses sesuai dengan perintah yang diberikan.

1. Operator Aritmatika

Operator aritmatika digunakan untuk melakukan operasi perhitungan matematika. Operator ini meliputi penjumlahan (+), pengurangan (-), perkalian (*), pembagian (/), pembagian bulat (//), modulus (%), dan perpangkatan (**).

2. Operator Assignment

Operator assignment digunakan untuk memberikan atau menetapkan nilai ke dalam sebuah variabel. Operator yang paling dasar adalah (=). Selain itu, terdapat assignment gabungan seperti +=, -=, *=, dan /= yang dapat digunakan untuk memperbarui nilai suatu variabel secara lebih sederhana.

3. Operator Bitwise

Operator bitwise digunakan untuk melakukan operasi pada data dalam bentuk bit atau representasi biner. Beberapa operator bitwise dalam Python yaitu AND (&), OR (|), XOR (^), NOT (~), left shift (<<), dan right shift (>>). Operator ini banyak digunakan dalam pengolahan data pada tingkat bit.

4. Operator Logika

Operator logika digunakan untuk menggabungkan atau memproses kondisi yang menghasilkan nilai benar (True) atau salah (False). Python memiliki tiga operator logika utama, yaitu **and**, **or**, dan **not**. Operator ini biasanya digunakan ketika sebuah program memiliki lebih dari satu kondisi.

5. Operator Perbandingan

Operator perbandingan digunakan untuk membandingkan dua nilai. Hasil dari perbandingan berupa nilai Boolean, yaitu True atau False. Operator yang digunakan antara lain sama dengan (==), tidak sama dengan (!=), lebih besar (>), lebih kecil (<), lebih besar atau sama dengan (>=), dan lebih kecil atau sama dengan (<=).

Dengan memahami berbagai jenis operator tersebut, programmer dapat membuat program yang mampu melakukan perhitungan, mengolah nilai variabel, membandingkan data, serta menentukan kondisi berdasarkan kebutuhan program.

### 2.2.1 Program "Aritmatika_2611537001"

<div style="font-family:'Times New Roman', Times, serif; font-size: 18px;">

Program aritmatika digunakan untuk melakukan operasi perhitungan matematika pada Python.
Pada program ini digunakan beberapa operator seperti penjumlahan (+), pengurangan (-), perkalian (*), dan pembagian (/).
Selain itu, terdapat operator modulus (%) untuk mendapatkan sisa pembagian dan perpangkatan (**) untuk menghitung pangkat.
Nilai yang digunakan dalam operasi dapat disimpan terlebih dahulu ke dalam variabel.
Setiap operasi akan menghasilkan nilai yang dapat ditampilkan menggunakan fungsi `print()`.
Program ini membantu memahami penerapan operator aritmatika dalam melakukan perhitungan sederhana menggunakan Python.


```python
#angka1_7001
#program ini menggunakan fungsi input ()
#nilai yang dimasukan akan di konversi menjadi tipe data integer

angka1 = int(input("input angka-1:"))
angka2 = int(input("input angka-2: "))

#penjumlahan
hasil = angka1 + angka2
print("\nOperasi Penjumlahan")
print("hasil =", hasil)

#pengurangan
hasil = angka1 - angka2
print("\nOperasi Pengurangan")
print("hasil =", hasil)

#Perkalian
hasil = angka1 * angka2
print("\nOperasi Perkalian")        
print("hasil =", hasil)

#pembagian, pembagian bulat, dan sisa bagi
if angka2 != 0:
    hasil = angka1 / angka2
    print("\nOperasi Pembagian")
    print("hasil =", hasil)

    hasil = angka1 // angka2
    print("\nOperasi Pembagian Bulat")
    print("hasil =", hasil)

    hasil = angka1 % angka2
    print("\nOperasi Sisa Bagi")
    print("hasil =", hasil)
else:
    print("Angka kedua tidak boleh bernilai 0.")

    #pangkat
hasil = angka1 ** angka2
print("\nOperasi Pangkat")
print("hasil =", hasil)
```

3. Pengurangan
Operator `-` digunakan untuk melakukan pengurangan antara `angka1` dan `angka2`. Hasil operasi kemudian disimpan dalam variabel `hasil` dan ditampilkan ke layar.

4. Perkalian
Operator `*` digunakan untuk mengalikan nilai `angka1` dengan `angka2`. Hasil perkalian disimpan pada variabel `hasil` dan kemudian ditampilkan menggunakan `print()`.

5. Pembagian
Operator `/` digunakan untuk melakukan pembagian antara `angka1` dan `angka2`. Program menggunakan kondisi `if angka2 != 0` untuk memastikan angka kedua tidak bernilai nol karena pembagian dengan nol tidak diperbolehkan.

Operator `//` digunakan untuk melakukan pembagian bulat. Hasil yang diperoleh berupa hasil pembagian tanpa bagian desimal.

### 2.2.2 Program "Assignment_2611537001"

Program assignment digunakan untuk memberikan atau mengubah nilai yang tersimpan dalam sebuah variabel.
Operator assignment dasar yang digunakan adalah `=`, yaitu untuk memberikan nilai awal pada variabel.
Selain itu, terdapat operator assignment gabungan seperti `+=`, `-=`, `*=`, dan `/=` untuk mengubah nilai variabel.
Operator tersebut memungkinkan proses perhitungan dan pemberian nilai dilakukan secara lebih sederhana.
Hasil dari setiap proses assignment dapat disimpan kembali ke dalam variabel yang sama.
Program ini membantu memahami cara mengelola dan memperbarui nilai variabel dalam Python.


```python
#angka1_7001
#program ini menggunakan fungsi input()
#nilai yang dimasukan akan di konversi menjadi tipe data interger
#program operator assignment dalam python

angka1_7001 =int(input("input angka-1:"))
angka2_7001 =int(input("input angka-2:"))

print("\nNilai awal angka1_7001 =", angka1_7001)
print("nilai angka2 =", angka2_7001)

#assignment biasa
hasil_7001 = angka1_7001
print("\nAssignment biasa (=")
print("hasil_7001 =", hasil_7001)

#assignment penambahan
hasil_7001 = angka1_7001
hasil_7001 += angka2_7001
print("\nAssignment menambahan (+=)")
print("hasil_7001 =", hasil_7001)

#assignment pengurangan 
hasil_7001 = angka1_7001
hasil_7001 -= angka2_7001
print("\nAssigment penambahan (-=)")
print("hasil_7001 =", hasil_7001)

#assignment perkalian
hasil_7001= angka1_7001
hasil_7001 *= angka2_7001
print("\nAssigment perkalian (*=)")
print("hasil_7001 =", hasil_7001)

#assigment pembagian, pembagian bulat, dan sisa bagi
if angka2_7001 != 0:
    hasil_7001 = angka1_7001
    hasil_7001 /= angka2_7001
    print("\nAssigment pembagian (/=)")
    print("hasil_7001 =", hasil_7001)
    #operator tambahan  
    hasil_7001 = angka1_7001
    hasil_7001 //= angka2_7001
    print("\nAssigment pembagian bulat (//=)")
    print("hasil_7001 =", hasil_7001)
    hasil_7001 = angka1_7001
    hasil_7001 %= angka2_7001
    print("\nAssigment sisa bagi (%=)")
    print("hasil_7001 =", hasil_7001)
else:
    print("\nPembagian tidak dapat dilakukan")
    print("angka kedua tidak boleh bernilai 0")

#operator tambahan: assignment perpangkatan
hasil_7001 = angka1_7001
hasil_7001 **= angka2_7001
print("\nAssigment berpangkatan (**=)")
print("hasil_7001 =", hasil_7001)
```

**Penjelasan:** 

1. Deklarasi dan Input Variabel
Bagian awal program digunakan untuk menerima dua nilai dari pengguna melalui fungsi input(). Nilai yang dimasukkan kemudian dikonversi menjadi tipe data integer menggunakan int() dan disimpan ke dalam variabel angka1_7001 dan angka2_7001.

2. Menampilkan Nilai Awal
Program menampilkan nilai awal dari kedua variabel menggunakan fungsi print(). Bagian ini bertujuan untuk menunjukkan nilai yang akan digunakan dalam berbagai operasi assignment.

3. Assignment Biasa (=)
Operator = digunakan untuk memberikan nilai angka1_7001 kepada variabel hasil_7001. Nilai tersebut kemudian ditampilkan sebagai hasil dari assignment biasa.

4. Assignment Penambahan (+=)
Pada bagian ini, hasil_7001 terlebih dahulu diberi nilai angka1_7001. Operator += kemudian digunakan untuk menambahkan nilai angka2_7001 ke dalam hasil_7001, sehingga hasil akhirnya merupakan penjumlahan kedua angka tersebut.

5. Assignment Pengurangan (-=)
Operator -= digunakan untuk mengurangi nilai hasil_7001 dengan angka2_7001. Sebelum operasi dilakukan, hasil_7001 diisi dengan nilai angka1_7001, sehingga hasil akhirnya merupakan pengurangan antara kedua angka.

6. Assignment Perkalian (*=)
Pada bagian ini digunakan operator *= untuk mengalikan nilai hasil_7001 dengan angka2_7001. Nilai awal hasil_7001 berasal dari angka1_7001, kemudian hasil perkalian ditampilkan.

7. Assignment Pembagian (/=)
Program menggunakan kondisi if angka2_7001 != 0 untuk memastikan nilai angka kedua tidak sama dengan nol. Jika memenuhi kondisi, operator /= digunakan untuk membagi nilai hasil_7001 dengan angka2_7001.

8. Assignment Pembagian Bulat dan Sisa Bagi (//= dan %=)
Operator //= digunakan untuk melakukan pembagian bulat, sedangkan %= digunakan untuk mendapatkan sisa hasil pembagian. Kedua operasi tersebut hanya dilakukan jika angka2_7001 tidak bernilai nol.

9. Kondisi Jika Angka Kedua Bernilai Nol
Bagian else dijalankan apabila angka2_7001 bernilai nol. Program kemudian menampilkan pesan bahwa operasi pembagian tidak dapat dilakukan karena angka kedua tidak boleh bernilai nol.

10. Assignment Perpangkatan (**=)
Pada bagian terakhir digunakan operator **= untuk melakukan perpangkatan. Nilai angka1_7001 digunakan sebagai nilai awal, kemudian dipangkatkan dengan angka2_7001 dan hasilnya disimpan dalam hasil_7001.

### 2.2.3 Program "Bitwise_2611537001"

Program bitwise digunakan untuk melakukan operasi pada bilangan dalam bentuk bit atau biner.
Pada program ini digunakan beberapa operator bitwise seperti &, |, ^, dan ~.
Operator & digunakan untuk operasi AND, sedangkan | digunakan untuk operasi OR.
Operator ^ digunakan untuk operasi XOR dan ~ digunakan untuk membalik nilai bit.
Selain itu, terdapat operator << dan >> yang digunakan untuk menggeser posisi bit ke kiri atau kanan.
Program ini membantu memahami cara kerja operasi bitwise pada bilangan menggunakan Python.

```python
#angka1_7001
# Program ini menggunakan fungsi input()

print("========================================")
print("3. OPERATOR BITWISE")
print("========================================")

angka1_7001 = int(input("Masukkan angka bitwise-1: "))
angka2_7001 = int(input("Masukkan angka bitwise-2: "))

print("Angka dalam bentuk desimal dan biner")
print("angka1 =", angka1_7001, "| biner =", bin(angka1_7001))
print("angka2 =", angka2_7001, "| biner =", bin(angka2_7001))

# Bitwise AND
hasil_7001 = angka1_7001 & angka2_7001
print("\nBitwise AND (&)")
print(angka1_7001, "&", angka2_7001, "=", hasil_7001)
print("Biner hasil =", bin(hasil_7001))
print("Biner hasil (8 bit) =", format(hasil_7001, "08b"))

# Bitwise OR
hasil_7001 = angka1_7001 | angka2_7001
print("\nBitwise OR (|)")
print(angka1_7001, "|", angka2_7001, "=", hasil_7001)
print("Biner hasil =", bin(hasil_7001))
print("Biner hasil (8 bit) =", format(hasil_7001, "08b"))

# Bitwise XOR
hasil_1017 = angka1_7001 ^ angka2_7001
print("\nBitwise XOR (^)")
print(angka1_7001, "^", angka2_7001, "=", hasil_7001)
print("Biner hasil =", bin(hasil_7001))
print("Biner hasil (8 bit) =", format(hasil_7001, "08b"))

# Bitwise NOT
hasil_1017 = ~angka1_7001
print("\nBitwise NOT (~)")
print("~", angka1_7001, "=", hasil_7001)
print("Biner hasil =", bin(hasil_7001))
print("Biner hasil (8 bit) =", format(hasil_7001, "08b"))

# Bitwise geser kiri
jumlah_geser_7001 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_7001 = angka1_7001 << jumlah_geser_7001
print("\nBitwise geser kiri (<<)")
print(angka1_7001, "<<", jumlah_geser_7001, "=", hasil_7001)
print("Biner hasil =", bin(hasil_7001))
print("Biner hasil (8 bit) =", format(hasil_7001, "08b"))

# Bitwise geser kanan
hasil_7001 = angka1_7001 >> jumlah_geser_7001
print("\nBitwise geser kanan (>>)")
print(angka1_7001, ">>", jumlah_geser_7001, "=", hasil_7001)
print("Biner hasil =", bin(hasil_7001))
print("Biner hasil (8 bit) =", format(hasil_1017, "08b"))
```

**Penjelasan:**

1. Judul dan Tampilan Program
Bagian awal program menggunakan print() untuk menampilkan judul Operator Bitwise sebagai penanda bahwa program akan melakukan operasi bitwise.

2. Input Variabel
Program meminta pengguna memasukkan dua angka menggunakan fungsi input(). Nilai yang dimasukkan dikonversi menjadi tipe data integer menggunakan int() dan disimpan ke dalam variabel angka1_7001 dan angka2_7001.

3. Menampilkan Bentuk Desimal dan Biner
Fungsi print() digunakan untuk menampilkan nilai kedua angka dalam bentuk desimal dan biner. Fungsi bin() digunakan untuk mengubah angka desimal menjadi representasi bilangan biner.

4. Bitwise AND (&)
Operator & digunakan untuk melakukan operasi AND pada setiap bit dari angka1_7001 dan angka2_7001. Hasil operasi kemudian disimpan dalam variabel hasil_7001 dan ditampilkan dalam bentuk desimal serta biner.

5. Bitwise OR (|)
Operator | digunakan untuk melakukan operasi OR pada setiap bit dari kedua angka. Hasil operasi disimpan ke dalam hasil_7001, kemudian ditampilkan dalam bentuk desimal dan biner.

6. Bitwise XOR (^)
Operator ^ digunakan untuk melakukan operasi XOR pada setiap bit dari kedua angka. Hasil operasi kemudian ditampilkan dalam bentuk desimal, biner, dan format biner 8 bit menggunakan format().

7. Bitwise NOT (~)
Operator ~ digunakan untuk membalik setiap bit pada nilai angka1_7001. Hasil operasi NOT kemudian disimpan dalam variabel hasil dan ditampilkan dalam bentuk desimal serta biner.

8. Input Jumlah Pergeseran Bit
Program meminta pengguna memasukkan jumlah pergeseran bit menggunakan input(). Nilai tersebut dikonversi menjadi integer dan disimpan dalam variabel jumlah_geser_7001.

9. Bitwise Geser Kiri (<<)
Operator << digunakan untuk menggeser bit pada angka1_7001 ke arah kiri sesuai jumlah pergeseran yang diberikan. Hasilnya kemudian ditampilkan dalam bentuk desimal dan biner.

10. Bitwise Geser Kanan (>>)
Operator >> digunakan untuk menggeser bit pada angka1_7001 ke arah kanan sesuai jumlah pergeseran yang diberikan. Hasil operasi kemudian ditampilkan dalam bentuk desimal dan biner 8 bit.

### 2.2.4 Program "logika_2611537001"

Program logika digunakan untuk mengolah suatu kondisi yang menghasilkan nilai True atau False.
Pada program ini digunakan beberapa operator logika seperti and, or, dan not.
Operator and menghasilkan True jika kedua kondisi bernilai benar.
Operator or menghasilkan True jika salah satu atau kedua kondisi bernilai benar.
Operator not digunakan untuk membalik nilai logika, dari True menjadi False atau sebaliknya.
Program ini membantu memahami penggunaan operator logika dalam menentukan dan menggabungkan suatu kondisi pada Python.

```python
#a1_7001
#program ini menggunakan fungsi input()
#program operator logika dalam python

#memasukan nilai boolean
#input tidak peka terhadap huruf besar dan kecil
a1_7001 = input("input nilai boolean-1 (True/False): ").strip().lower() == "true"
a2_7001 = input("input nilai boolean-2 (True/False): ").strip().lower() == "true"

print("\nA1_7001:", a1_7001)
print("A2_7001:", a2_7001)

#konjungsi: bernilai True jika kedua operand bernilai True
hasil = a1_7001 and a2_7001
print("\nKonjungsi (AND)")
print("A1_7001 and A2_7001 =", hasil)

#disjungsi: bernilai True jika salah satunya true
hasil = a1_7001 or a2_7001
print("\nDisjungsi (OR)")
print("A1_7001 or A2_7001 =", hasil)

#negasi A1_7001: membalik nilai A1_7001
hasil = not a1_7001
print("\nNegasi A1_7001 (NOT)")
print("not A1_7001 =", hasil)

#negasi A2_7001: membalikan nilai A2_7001 
hasil = not a2_7001
print("\nNegasi A2_7001 (NOT)")
print("not A2_7001 =", hasil)

#XOR: bernilai true jika kedua bilangan berbeda
hasil = a1_7001 != a2_7001
print("\nDiskonjungsi ekslusif (XOR)")
print("A1_7001 XOR A2_7001 =", hasil)
```

**Penjelasan:**

1. Deklarasi Program
Komentar pada bagian awal menjelaskan bahwa program menggunakan fungsi input() dan membahas penggunaan operator logika dalam Python.

2. Input Nilai Boolean
Variabel a1_7001 dan a2_7001 digunakan untuk menyimpan dua nilai Boolean yang dimasukkan oleh pengguna. Fungsi .strip() digunakan untuk menghilangkan spasi pada input, sedangkan .lower() mengubah huruf menjadi huruf kecil. Kemudian == "true" digunakan untuk menentukan apakah input bernilai True atau False.

3. Menampilkan Nilai Boolean
Fungsi print() digunakan untuk menampilkan nilai yang telah dimasukkan dan disimpan pada variabel a1_7001 dan a2_7001.

4. Konjungsi (AND)
Operator and digunakan untuk menggabungkan dua kondisi. Hasilnya bernilai True apabila kedua nilai Boolean bernilai True.

5. Disjungsi (OR)

### 2.2.5 Program "nama lainnya_2611537001"

Operator keanggotaan digunakan untuk memeriksa apakah suatu nilai terdapat di dalam sebuah data atau kumpulan data.
Pada program ini digunakan operator in dan not in untuk melakukan pengecekan keanggotaan.
Operator in menghasilkan nilai True jika suatu nilai ditemukan di dalam data yang diperiksa.
Operator not in menghasilkan nilai True jika suatu nilai tidak ditemukan di dalam data.
Hasil dari pengecekan operator keanggotaan berupa nilai Boolean, yaitu True atau False.
Program ini membantu memahami cara melakukan pengecekan keberadaan suatu nilai dalam data menggunakan Python.

```python
#angka1_7001
#program ini menggunakan fungsi input()
#program operator keanggotaan dan identitas
print("===================================")
print("1. OPERATOR KEANGGOTAAN")
print("===================================")

#input beberapa data yang dipisahkan dengan koma
input_data = input ("masukan beberapa angka, pisahkan dengan koma: ")

#mengubah input menjadi list integer 
data = [int(angka.strip()) for angka in input_data.split(",")]

nilai_dicari = int(input("memasukan angka yang ingin dicari: "))

#operator in
hasil_7001 = nilai_dicari in data
print("\nOperator keanggotaan IN")
print(nilai_dicari, "in", data, "=", hasil_7001)

#operator not in
hasil_7001 = nilai_dicari not in data
print("\nOperator keanggotaan NOT IN")
print(nilai_dicari, "not in", data, "=", hasil_7001)

print("===================================")
print("2. OPERATOR IDENTITAS")
print("===================================")

# objek1 menggunakan list dari input 
objek1 = data

#objek2 merujuk pada objek yang sama dengan objek1
objek2 = objek1

#objek3 memiliki isi yang sama, tetaapi merupaan objek terbaru
objek3 = data.copy()

print("objek1 =", objek1)
print("objek2 =", objek2)
print("objek3 =", objek3)

#operator is
hasil_7001 = objek1 is objek2
print("\nOperator idetitas IS")
print("objek1 is objek2 =", hasil_7001)

#operator is not
hasil_7001 = objek1 is not objek3
print("\nOperator idetitas IS NOT")
print("objek1 is not objek3 =", hasil_7001)

#membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai")
print("objek1 is objek3 =", objek1 is objek3)
print("objek1 == objek3", objek1 == objek3)
```

### 2.2.6 Program "perbandingan_2611537001"

Operator perbandingan digunakan untuk membandingkan dua nilai dalam suatu program.
Operator yang digunakan antara lain ==, !=, >, <, >=, dan <=.
Operator == digunakan untuk memeriksa apakah dua nilai sama, sedangkan != untuk memeriksa apakah nilainya berbeda.
Operator >, <, >=, dan <= digunakan untuk membandingkan besar atau kecilnya dua nilai.
Hasil dari setiap perbandingan berupa nilai Boolean, yaitu True atau False.
Operator perbandingan dapat digunakan untuk menentukan kondisi dan membantu program dalam mengambil keputusan.


```python
#angka1_7001
#program ini menggunakan fungsi input()
#nilai yang dimasukan akan di konversi menjadi tipe data integer
#program operatoer perbandingan dalam python

angka1_7001 = int(input("input angka-1:"))
angka2_7001 = int(input("input angka-2: ")) 

# lebih besar dari
hasil = angka1_7001 > angka2_7001
print("\nOperasi Lebih Besar Dari")
print("angka1_7001 > angka2_7001:", hasil)

# lebih kecil dari
hasil = angka1_7001 < angka2_7001
print("\nOperasi Lebih Kecil Dari")
print("angka1_7001 < angka2_7001:", hasil)

# lebih besar dari atausama dengan
hasil = angka1_7001 >= angka2_7001
print("\nOperasi Lebih Besar Dari Sama Dengan")
print("angka1_7001 >= angka2_7001:", hasil)

# lebih kecil dari atau sama dengan
hasil = angka1_7001 <= angka2_7001
print("\nOperasi Lebih Kecil Dari Sama Dengan")
print("angka1_7001 <= angka2_7001:", hasil)   

# sama dengan
hasil = angka1_7001 == angka2_7001
print("\nOperasi Sama Dengan")
print("angka1_7001 == angka2_7001:", hasil)

# tidak sama dengan
hasil = angka1_7001 != angka2_7001
print("\nOperasi Tidak Sama Dengan")
print("angka1_7001 != angka2_7001:", hasil)

#tambahan: perbandingan berantai dalam python
hasil = 0 < angka1_7001 < 100
print("\nPerbandingan Berantai")
print("0 < angka1_7001 < 100:", hasil)

hasil = 0 < angka2_7001 < 100
print("0 < angka2_7001 < 100:", hasil)
```

**Penjelasan:**

1. Deklarasi dan Input Variabel

angka1_7001 = int(input("input angka-1:"))
angka2_7001 = int(input("input angka-2: "))

Digunakan untuk memasukkan dua angka dari pengguna. Fungsi input() menerima masukan, sedangkan int() mengubah masukan menjadi tipe data integer. Nilai tersebut disimpan ke dalam variabel angka1_7001 dan angka2_7001.

2. Perbandingan Lebih Besar Dari (>)

hasil = angka1_7001 > angka2_7001
print("\nOperasi Lebih Besar Dari")
print("angka1_7001 > angka2_7001:", hasil)

Operator > digunakan untuk membandingkan apakah nilai angka1_7001 lebih besar dari angka2_7001. Hasil perbandingan disimpan dalam variabel hasil, kemudian ditampilkan menggunakan print().

3. Perbandingan Lebih Kecil Dari (<)

hasil = angka1_7001 < angka2_7001
print("\nOperasi Lebih Kecil Dari")
print("angka1_7001 < angka2_7001:", hasil)

Operator < digunakan untuk mengetahui apakah nilai angka1_7001 lebih kecil dari angka2_7001. Hasil perbandingan berupa True atau False dan disimpan dalam variabel hasil.

4. Perbandingan Lebih Besar atau Sama Dengan (>=)

hasil = angka1_7001 >= angka2_7001
print("\nOperasi Lebih Besar Dari Sama Dengan")
print("angka1_7001 >= angka2_7001:", hasil)

Operator >= digunakan untuk memeriksa apakah angka1_7001 lebih besar atau memiliki nilai yang sama dengan angka2_7001. Hasilnya kemudian disimpan dan ditampilkan.

5. Perbandingan Lebih Kecil atau Sama Dengan (<=)

hasil = angka1_7001 <= angka2_7001
print("\nOperasi Lebih Kecil Dari Sama Dengan")
print("angka1_7001 <= angka2_7001:", hasil)

Operator <= digunakan untuk memeriksa apakah angka1_7001 lebih kecil atau sama dengan angka2_7001. Hasil perbandingan disimpan dalam hasil dan ditampilkan sebagai nilai Boolean.

6. Perbandingan Sama Dengan (==)

hasil = angka1_7001 == angka2_7001
print("\nOperasi Sama Dengan")
print("angka1_7001 == angka2_7001:", hasil)

Operator == digunakan untuk memeriksa apakah nilai angka1_7001 sama dengan angka2_7001. Jika kedua nilai sama, hasilnya True, sedangkan jika berbeda hasilnya False.

7. Perbandingan Tidak Sama Dengan (!=)

hasil = angka1_7001 != angka2_7001
print("\nOperasi Tidak Sama Dengan")
print("angka1_7001 != angka2_7001:", hasil)

Operator != digunakan untuk memeriksa apakah nilai angka1_7001 dan angka2_7001 berbeda. Jika nilainya berbeda, hasilnya True, sedangkan jika sama hasilnya False.

8. Perbandingan Berantai pada Angka Pertama

hasil = 0 < angka1_7001 < 100
print("\nPerbandingan Berantai")
print("0 < angka1_7001 < 100:", hasil)

Perbandingan berantai digunakan untuk memeriksa apakah nilai angka1_7001 lebih besar dari 0 dan lebih kecil dari 100 secara bersamaan. Hasil perbandingan disimpan dalam hasil dan kemudian ditampilkan.

9. Perbandingan Berantai pada Angka Kedua

hasil = 0 < angka2_7001 < 100
print("0 < angka2_7001 < 100:", hasil)

Bagian ini melakukan perbandingan berantai terhadap angka2_7001. Program memeriksa apakah nilai tersebut berada di antara 0 dan 100, kemudian menampilkan hasilnya dalam bentuk True atau False.

10. Hasil Perbandingan

Seluruh operasi perbandingan menghasilkan nilai Boolean, yaitu True atau False. Nilai tersebut menunjukkan apakah kondisi perbandingan yang dilakukan terpenuhi atau tidak.

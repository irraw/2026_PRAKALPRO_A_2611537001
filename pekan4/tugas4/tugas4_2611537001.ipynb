## 2.1 Dasar Teori

<div style="font-family:'Times New Roman', Times, serif; font-size: 18px;">

Struktur percabangan merupakan struktur dalam pemrograman yang digunakan untuk menentukan alur program berdasarkan kondisi tertentu. Dengan percabangan, program dapat menjalankan perintah yang berbeda sesuai dengan kondisi yang terpenuhi. Struktur ini memungkinkan program mengambil keputusan secara otomatis berdasarkan data atau kondisi yang diberikan. Dalam Python, struktur percabangan dapat digunakan untuk membuat program yang lebih dinamis dan mampu menangani berbagai kemungkinan kondisi.


1. Percabangan if

Percabangan if digunakan untuk menjalankan suatu perintah apabila kondisi yang diberikan bernilai benar (True). Jika kondisi bernilai salah (False), maka perintah di dalam if tidak akan dijalankan.

2. Percabangan if-else

Percabangan if-else digunakan ketika terdapat dua kemungkinan kondisi. Jika kondisi pada if bernilai benar, program akan menjalankan perintah pada bagian if. Sebaliknya, jika kondisi bernilai salah, program akan menjalankan perintah pada bagian else.

3. Percabangan if-elif-else

Percabangan if-elif-else digunakan untuk memeriksa beberapa kondisi secara berurutan. Program akan menjalankan perintah pada kondisi pertama yang bernilai benar. Jika tidak ada kondisi yang terpenuhi, maka bagian else akan dijalankan.

4. Percabangan Multi-If

Multi-if merupakan penggunaan beberapa pernyataan if secara terpisah dalam satu program. Setiap kondisi akan diperiksa secara independen, sehingga beberapa kondisi dapat terpenuhi dan dijalankan secara bersamaan.

5. Percabangan match-case

match-case digunakan untuk mencocokkan suatu nilai dengan beberapa pilihan atau pola tertentu. Struktur ini dapat digunakan untuk menangani banyak pilihan kondisi berdasarkan nilai yang diberikan.

### 2.2.5 Program "multi_if2_2611537001"

Program tersebut merupakan program yang menggunakan struktur multi-if untuk menghitung total diskon dan jumlah pembayaran berdasarkan total belanja, status member, dan kode promo. Setiap kondisi diperiksa secara terpisah sehingga diskon dapat ditambahkan atau diakumulasikan apabila pengguna memenuhi beberapa syarat sekaligus. Setelah seluruh kondisi diperiksa, program menghitung nominal diskon dan total pembayaran yang harus dibayar.

```python
#buat program untuk kondisional if
#total_belanja_7001
#program ini menggunakan fungsi input
#program menghitung diskon belanja

#input dari user
total_belanja_7001 = float(input("masukan total belanja (Rp) : "))

#input status member (mengecek apakag user mengetik 'y' atau 'ya') 
input_member_7001 = input("apakah anda member? (y/t) : ").strip( ).lower( )
is_member_7001 = input_member_7001 in ["y","ya"]

#input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_7001 = input("apakah kode promo valid? (y/t):").strip( ).lower( )
kode_promo_valid_7001 = input_promo_7001 in ["y","ya"]

total_diskon_persen_7001 = 0

#multi-if terpisah: setiap kondisi diperiksa secara independen
#diskon bisa di tumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_7001 > 1000000:
    total_diskon_persen_7001 += 10 # diskon belanja besar

if is_member_7001:
    total_diskon_persen_7001 += 5 # diskon member

if kode_promo_valid_7001:
    total_diskon_persen_7001 += 15 #diskon voucher

#menghitung nominal diskon dan total bayar
nominal_diskon_7001 = total_belanja_7001 * (total_diskon_persen_7001 / 100)
total_bayar_7001 = total_belanja_7001 - nominal_diskon_7001

#output hasil 
print("\n--- Rincian pembayaran ---")
print(f"total diskon : {total_diskon_persen_7001}% (Rp{nominal_diskon_7001:,.0f})")
print(f"total bayar  : Rp {total_bayar_7001:,.0f}")

print(f"total diskon yang anda dapatkan : {total_diskon_persen_7001}%")
#output: total diskon yang anda dapatkan: 30% jika belanja > 1 juta,member, dan kode promo valid
```

### 2.2.7 Program "match_case1"

Program tersebut merupakan program yang menggunakan struktur percabangan match-case untuk mengubah angka yang dimasukkan pengguna menjadi nama bulan. Pengguna memasukkan angka dari 1 sampai 12, kemudian program mencocokkan angka tersebut dengan setiap case yang tersedia. Jika angka sesuai dengan salah satu case, program menampilkan nama bulan yang sesuai. Jika angka yang dimasukkan tidak berada pada rentang 1 sampai 12, program akan menampilkan bahwa angka tersebut tidak valid.

```python
#buat program untuk match case
#ipk_7001
#program ini menggunakan fungsi input()
#program konversi angka menjadi nama bulan

bulan_7001 = int(input("masukan angka bulan (1-12):"))

match bulan_7001:
    case 1:
        print("januari")
    case 2:
        print("februari")
    case 3:
            print("maret")
    case 4:
            print("april")
    case 5:
            print("mei")
    case 6:
            print("juni")
    case 7:
            print("juli")
    case 8:
            print("agustus")
    case 9:
            print("september")
    case 10:
            print("oktober")
    case 11:
            print("november")
    case 12:
            print("desember")
    case _:
            print("angka tidak valid")
```

   Digunakan untuk memulai struktur match-case yang akan mencocokkan nilai dari variabel bulan_7001 dengan beberapa pilihan case.

   Digunakan untuk mencocokkan angka yang dimasukkan dengan angka 1 sampai 12. Setiap angka memiliki perintah untuk menampilkan nama bulan yang sesuai.

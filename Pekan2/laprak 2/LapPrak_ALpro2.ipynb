## 2.1 Dasar Teori

### 1. Variabel dan Tipe Data

Variabel dapat dipahami sebagai suatu penamaan yang digunakan untuk menyimpan sebuah data di dalam program. Data yang tersimpan melalui variabel dapat digunakan kembali ketika program melakukan proses tertentu. Pada Python, programmer tidak perlu menentukan tipe data variabel ketika membuatnya. Hal ini disebabkan Python menggunakan sistem *dynamically typed*, sehingga tipe data akan dikenali berdasarkan nilai yang diberikan kepada variabel.

Dalam praktikum ini, beberapa konsep dasar Python yang dipelajari meliputi tipe data numerik, string, boolean, penggunaan fungsi `input()`, serta konstanta. Masing-masing memiliki karakteristik dan kegunaan yang berbeda dalam pembuatan program.

### 2. Tipe Data Numerik

Data yang berupa angka dalam Python termasuk ke dalam kelompok tipe numerik. Tipe ini dapat digunakan untuk berbagai operasi perhitungan. Python menyediakan beberapa bentuk data numerik, di antaranya:

* **Integer (`int`)**, yaitu data berupa bilangan bulat, misalnya `100`.
* **Float**, yaitu data berupa angka yang memiliki nilai pecahan atau desimal, misalnya `10.2345`.
* **Complex**, yaitu bilangan yang memiliki komponen real dan imajiner. Bagian imajiner dituliskan menggunakan `j`, contohnya `100+3j`.

Selain digunakan untuk perhitungan, jenis data yang dimiliki suatu variabel juga dapat diketahui dengan menggunakan fungsi `type()`. Fungsi tersebut akan menampilkan kelas atau tipe dari nilai yang terdapat pada variabel.

### 3. Tipe Data String

String digunakan ketika data yang ingin disimpan berupa karakter atau teks. Contohnya dapat berupa nama, alamat, maupun kalimat. Penulisan string di Python dapat menggunakan tanda petik satu (`' '`), tanda petik dua (`" "`), atau tanda petik tiga (`""" """`). Penggunaan tanda petik tiga memungkinkan programmer menuliskan teks yang terdiri dari beberapa baris.

### 4. Tipe Data Boolean

Boolean merupakan tipe data yang berkaitan dengan kondisi logika. Nilai yang dimiliki boolean hanya terdiri atas `True` dan `False`. Tipe data ini banyak digunakan ketika program perlu menentukan hasil dari suatu perbandingan atau kondisi tertentu. Boolean juga memiliki peran penting dalam proses pengambilan keputusan, misalnya ketika digunakan bersama struktur percabangan `if`.

### 5. Fungsi `input()`

Python menyediakan fungsi `input()` untuk memungkinkan program menerima data yang dimasukkan oleh pengguna. Data yang diperoleh melalui fungsi tersebut secara default dianggap sebagai string. Oleh karena itu, apabila masukan pengguna berupa angka dan akan digunakan untuk perhitungan, diperlukan proses konversi tipe data terlebih dahulu.

Konversi tersebut dapat dilakukan menggunakan fungsi seperti `int()` untuk mengubah data menjadi bilangan bulat atau `float()` untuk mengubahnya menjadi bilangan desimal. Dengan demikian, data hasil masukan dapat diproses sesuai dengan kebutuhan program.

### 6. Konstanta

Konstanta merupakan nilai yang secara konsep ditetapkan untuk tetap sama selama program digunakan. Salah satu contoh yang umum adalah nilai PI yang digunakan dalam berbagai perhitungan matematika. Berbeda dengan beberapa bahasa pemrograman lain, Python tidak memiliki aturan khusus yang secara otomatis membuat suatu variabel menjadi tidak dapat diubah.

Sebagai gantinya, programmer biasanya menggunakan konvensi penamaan dengan huruf kapital untuk menunjukkan bahwa sebuah variabel dimaksudkan sebagai konstanta, misalnya `PI`. Python juga menyediakan `Final` dari modul `typing` sebagai penanda bahwa suatu nilai seharusnya tidak mengalami perubahan. Namun, penanda tersebut tidak secara langsung mencegah perubahan ketika program dijalankan oleh interpreter; peringatannya dapat diberikan oleh *type checker*.

### 2.2.1 Numerik "A_2611537001"


```python
# Program pertama
#create a variable with integer value
a_7001=100
print("the type of variable having value", a_7001, "is", type(a_7001))

#create a variable with float value
b=10.2345
print("the type of variable having value", b,"is", type(b))

#create a variable with complex value
c=100+3j
print("the type of variable having value", c, "is", type(c))
```

### 2.2.3 Boolean "A_2611537001"

```python
#Nilai_7001
#deklarasi variabel dengan tipe data boolean
is_lulus = True
is_cumlaude = True

#menggunakan boolean
nilai = 85
batas_lulus = 75

# Menentukan nilai boolean dari kondisi
status_kelulusan = nilai >= batas_lulus #Hasilnya akan true

print ("=== Check kelulusan ===")
print("Nilai:", nilai)
print("Apakah lulus?", status_kelulusan)
if is_lulus and is_cumlaude:
    print("Selamat, Anda lulus dengan predikat cumlaude!")
```

### 2.2.4 Input "A_2611537001"

```python
#panjang_7001
#program ini menggunakan fungsi input () untuk meminta pengguna
#memasukkan panjang dan lebar dari sebuah persegi panjang
#nilai yang dimasukan akan dikonversi menjadi tipe data integer
panjang_7001 = input('masukan nilai panjang: ')
lebar_7001 = input('masukan nilai lebar: ')
luas_7001 = int(panjang_7001) * int(lebar_7001)
print("luas =", luas_7001)
```

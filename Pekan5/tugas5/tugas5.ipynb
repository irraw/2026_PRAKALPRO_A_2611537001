## 2.1 Dasar Teori

<div style="font-family:'Times New Roman', Times, serif; font-size: 18px;">

Sebelum masuk ke praktik, penulis perlu memahami teori dasar yang melandasi materi pekan ini. Pembahasan berikut mencakup konsep perulangan secara umum, perulangan for beserta fungsi range() yang menyertainya, serta nested loop sebagai pengembangan dari perulangan tunggal. Pemahaman terhadap teori ini menjadi bekal penting agar kode program yang dibuat pada praktikum dapat dipahami alur kerjanya, bukan sekadar disalin dan dijalankan.


1. perulangan

Perulangan atau looping adalah struktur dalam pemrograman yang digunakan untuk menjalankan suatu blok kode secara berulang selama kondisi tertentu terpenuhi atau sebanyak jumlah yang telah ditentukan. Dengan perulangan, instruksi yang sama tidak perlu ditulis berkali-kali sehingga kode menjadi lebih singkat, rapi, dan mudah diperbaiki. Pada Python, terdapat dua jenis perulangan utama, yaitu for dan while. Pada praktikum ini, pembahasan difokuskan pada perulangan for.

2. Perulangan for 

Perulangan for pada Python digunakan untuk melakukan iterasi terhadap suatu urutan data (sequence), seperti list, string, tuple, atau rentang angka. Pada setiap putaran, variabel perulangan akan menyimpan satu elemen dari urutan tersebut, lalu blok kode di dalamnya dijalankan. Perulangan berhenti secara otomatis ketika seluruh elemen sudah diproses. Berbeda dengan bahasa seperti C atau Java, for di Python tidak memerlukan inisialisasi, kondisi, dan penambahan nilai secara eksplisit.

3. Fungsi range()

Fungsi range() digunakan untuk menghasilkan deretan bilangan bulat yang biasa dipadukan dengan perulangan for. Fungsi ini memiliki tiga bentuk penulisan:

range(stop): menghasilkan angka dari 0 sampai stop - 1.
range(start, stop): menghasilkan angka dari start sampai stop - 1.
range(start, stop, step): menghasilkan angka dari start sampai sebelum stop dengan selisih sebesar step.

4. Nested loop

Nested loop adalah perulangan yang berada di dalam perulangan lainnya. Perulangan yang berada di luar disebut perulangan luar (outer loop), sedangkan perulangan yang berada di dalam disebut perulangan dalam (inner loop). Setiap kali perulangan luar berputar satu kali, perulangan dalam akan berputar sampai selesai terlebih dahulu. Dengan demikian, jumlah total putaran sama dengan jumlah putaran perulangan luar dikalikan jumlah putaran perulangan dalam.

### 2.2.1 Program "nested_for1_2611537001"

<div style="font-family:'Times New Roman', Times, serif; font-size: 18px;">

Program ini membuat pola angka dengan titik di sebelah kirinya. Baris pertama punya titik paling banyak (batas dikurangi 1), lalu jumlah titik berkurang satu di setiap baris, sampai baris terakhir yang hanya berisi angka tanpa titik. Perulangan luar mengatur nomor baris, perulangan dalam mengatur jumlah titik.

```python
#ulang_7001
#program ini menggunakan fungsi input()

batas_7001 = int(input("masukan nilai batas: "))
for line in range (1, batas_7001 + 1) :
    for j_7001 in range(1, (-1 * line + batas_7001) + 1):
        print (".", end=" ")
    print(line)
```

**Penjelasan:**


1. Komentar di awal program. Dua baris pertama hanya catatan untuk pembaca dan tidak dijalankan oleh Python.

2. Input nilai batas. Program meminta pengguna mengetik angka, diubah dengan int() menjadi bilangan bulat, lalu disimpan di variabel batas_7001.

3. Perulangan luar (baris). Perulangan for dengan variabel line berjalan dari 1 sampai batas_7001. Setiap putaran luar mewakili satu baris output.

4. Perulangan dalam (titik). Perulangan for dengan variabel j_7001 berjalan dari 1 sampai batas_7001 dikurangi line. Artinya jumlah titik di setiap baris berkurang seiring bertambahnya nomor baris. Ini yang disebut nested loop.

5. Mencetak titik. Di setiap putaran dalam, program mencetak tanda "." dengan end=" " sehingga titik-titik tampil mendatar dengan pemisah spasi.

6. Mencetak nomor baris. Perintah print(line) sejajar dengan perulangan dalam, jadi dijalankan setelah semua titik di baris itu selesai dicetak. Program menampilkan nilai line lalu pindah ke baris baru.

### 2.2.4 Program "nested_for4_2611537001"

Program ini membuat pola persegi panjang berbingkai dengan tinggi sesuai input genap dan lebar 2 x tinggi - 2. Bingkai atas dan bawah memakai "#" dan "=", sisi kiri dan kanan memakai "|", sedangkan bagian dalamnya berisi pasangan tanda "<" dan ">" yang bergerak mendekati tepi lalu kembali ke arah tengah di setiap baris, dengan titik "." mengisi ruang di antara pasangan tersebut. Perulangan luar mengatur baris, perulangan dalam mengatur kolom.

```python
# ulang_1234
# Program ini menggunakan fungsi input()

tinggi_7001 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_7001 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_7001 = tinggi_7001
    c_7001 = a_7001
    lebar_7001 = (2 * tinggi_7001) - 2

    for i in range(1, tinggi_7001 + 1):
        b_7001 = c_7001 + 1

        for j in range(1, lebar_7001 + 1):

            # Baris atas dan bawah
            if i == 1 or i == tinggi_7001:
                if j == 1 or j == lebar_7001:
                    print("#", end="")
                else:
                    print("=", end="")
            # Baris isi
            else:
                if j == 1 or j == lebar_7001:
                    print("|", end="")
                else:
                    if j == c_7001:
                        print("<", end="")
                    elif j == b_7001:
                        print(">", end="")
                    elif j == (lebar_7001 - c_7001):
                        print("<", end="")
                    elif j == (lebar_7001 - c_7001 + 1):
                        print(">", end="")
                    elif j > b_7001 and j < (lebar_7001 - c_7001):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # Logika asli Java
        a_7001 -= 2

        if a_7001 <= 0:
            c_7001 = (-a_7001) + 2
        else:
            c_7001 = a_7001
```

**Penjelasan:**

4. Menyiapkan variabel. a_7001 dan c_7001 diisi sama dengan tinggi_7001, sedangkan lebar_7001 dihitung dengan rumus (2 x tinggi) - 2. Variabel c_7001 menentukan posisi kolom tanda "<", dan a_7001 dipakai untuk mengatur pergeseran posisi tersebut.

5. Perulangan luar (baris). Perulangan for dengan variabel i berjalan dari 1 sampai tinggi_7001, mewakili baris pola. Di awal setiap baris, b_7001 diisi c_7001 + 1 sebagai posisi tanda ">" yang berpasangan dengan tanda "<".

6. Perulangan dalam (kolom). Perulangan for dengan variabel j berjalan dari 1 sampai lebar_7001, mewakili kolom pada setiap baris. Inilah yang disebut nested loop.

Sisi kanan membentuk pasangan simetris, yaitu "<" di kolom lebar_7001 - c_7001 dan ">" di kolom setelahnya. Kolom yang berada di antara pasangan kiri dan kanan dicetak ".", sedangkan sisanya dicetak spasi.

10. Pindah baris. Perintah print() setelah perulangan dalam sejajar dengan perulangan dalam, sehingga dijalankan setiap satu baris selesai untuk pindah ke baris berikutnya.

### 2.2.7 Program "perulangan_for3_2611537001 "

Program ini meminta pengguna memasukkan sebuah angka, lalu menampilkan deretan angka dari 1 sampai angka tersebut dengan tanda + di antaranya. Variabel jumlah_7001 bertambah 1 di setiap putaran, sehingga setelah perulangan selesai nilainya sama dengan banyaknya putaran. Pada putaran terakhir, program mencetak tanda = diikuti nilai jumlah_7001, lalu di akhir menampilkan hasil "jumlah =" sekali lagi.

```python
#ulang_7001
#program ini menggunakan fungsi input()

ulang_7001 = int(input("masukan jumlah perulangan: "))

jumlah_7001 = 0
for i_7001 in range(1, ulang_7001 + 1):
    print(i_7001, end= " ")
    jumlah_7001 = jumlah_7001 +1

    if i_7001 < ulang_7001:
     print("+", end=" ") 
    else:
      print(" = ",jumlah_7001, end=" ")
print()
print("jumlah =", jumlah_7001)
```

**Penjelasan:**

### 2.2.8 Program "tugas_2611537001"

Program ini membuat pola segitiga bintang rata tengah (piramida). Perulangan luar mengatur baris, perulangan dalam pertama mencetak spasi pengisi di sebelah kiri, dan perulangan dalam kedua mencetak bintang. Baris pertama punya 1 bintang, baris kedua 2 bintang, dan seterusnya sampai baris terakhir berisi bintang sebanyak tinggi.

```python
#tugas_7001

tinggi_7001 = int(input("masukan tinggi pola: "))

# baris spasi
for i_7001 in range (1, tinggi_7001 + 1) :
    for j_7001 in range(tinggi_7001 - i_7001) :
        print(" ", end="")

# baris titik
    for a in range(i_7001) :
        print("* ", end="")
    print()
```

**Penjelasan:**

1. Komentar di awal program. Baris pertama hanya catatan untuk pembaca dan tidak dijalankan oleh Python.

2. Input tinggi pola. Program meminta pengguna mengetik angka, diubah dengan int() menjadi bilangan bulat, lalu disimpan di variabel tinggi_7001.

3. Perulangan luar (baris). Perulangan for dengan variabel i_7001 berjalan dari 1 sampai tinggi_7001. Setiap putaran luar mewakili satu baris pola.

4. Perulangan dalam pertama (spasi). Perulangan for dengan variabel j_7001 berjalan sebanyak tinggi_7001 dikurangi i_7001 kali. Di setiap putaran, program mencetak satu spasi tanpa pindah baris. Jumlah spasi berkurang seiring bertambahnya nomor baris, sehingga bintang bergeser ke kiri di baris-baris berikutnya.

5. Perulangan dalam kedua (bintang). Perulangan for dengan variabel a berjalan sebanyak i_7001 kali. Di setiap putaran, program mencetak "* " (bintang dan spasi). Jumlah bintang bertambah satu di setiap baris. Kedua perulangan dalam ini bersarang di dalam perulangan luar, sehingga disebut nested loop.

6. Pindah baris. Perintah print() sejajar dengan perulangan dalam, jadi dijalankan setelah spasi dan bintang pada satu baris selesai dicetak, untuk pindah ke baris berikutnya.
